# **Aggregations, Axis Math & Linear Algebra**

Is phase me hum seekhenge ki bade arrays me se summary statistics (Sum, Mean, Min, Max) kaise nikaalte hain, row/column wise calculation kaise karte hain (`axis=0` vs `axis=1`), aur Machine Learning ka core math (Dot Product) kaise perform karte hain.

---

### **1. Basic Summary Statistics (Aggregations)**

Puri matrix ya array ki ek single consolidated value nikalna:

* `np.sum(arr)` : Sabhi numbers ka total sum.
* `np.mean(arr)` : Average (Ausat) calculate karna.
* `np.std(arr)` : Standard Deviation (Data kitna faila hua hai).
* `np.min(arr)` / `np.max(arr)` : Minimum aur Maximum value nikalna.

```python
import numpy as np

data = np.array([12, 45, 67, 89, 23, 56])
print("Data:", data)

print("\n--- Summary Statistics ---")
print("Total Sum:", np.sum(data))
print("Mean (Average):", np.mean(data))
print("Standard Deviation:", np.std(data))
print("Minimum Value:", np.min(data))
print("Maximum Value:", np.max(data))
```

### **2. Axis-Based Math (`axis=0` vs `axis=1`)**

2D Table (Matrix) me calculation karte waqt direction define karna mandatory hota hai:

* **`axis=0` (Downwards / Columns ke along):** 
  - Har ek column ke elements par vertically calculation hoti hai.
  - Output har column ka ek answer deta hai.
* **`axis=1` (Across / Rows ke along):** 
  - Har ek row ke elements par horizontally calculation hoti hai.
  - Output har row ka ek answer deta hai.

> **Mental Trick:** `axis=0` collapse rows (columns bachte hain), `axis=1` collapse columns (rows bachti hain).

```python
# 3 Students ke 3 Subjects ke marks (3x3 Matrix)
scores = np.array([
    [80, 85, 90],  # Student 0
    [60, 70, 75],  # Student 1
    [95, 90, 85]   # Student 2
])

print("Marks Matrix (3 Students x 3 Subjects):\n", scores)

# Column-wise sum (axis=0): Har subject ka total
print("\nHar Subject ka Total (axis=0):", np.sum(scores, axis=0))

# Row-wise mean (axis=1): Har student ka average percentage
print("Har Student ka Average (axis=1):", np.mean(scores, axis=1))
```

### **3. Position Finders: `argmin()` & `argmax()`**

Yeh functions values ke bajaye us value ka **Index (Position)** return karte hain.

* **`np.argmax(arr)`:** Sabse badi value kis index par baithi hai.
* **`np.argmin(arr)`:** Sabse choti value kis index par baithi hai.

* **ML Relevance:** Classification models me probabilities di hoti hain (jaise `[0.10, 0.85, 0.05]`); `np.argmax()` batata hai ki prediction class `1` hai.

```python
# Model predictions for 4 classes: Cat, Dog, Bird, Fish
class_probabilities = np.array([0.05, 0.78, 0.12, 0.05])
classes = ["Cat", "Dog", "Bird", "Fish"]

print("Probabilities:", class_probabilities)

# Sabse badi probability ka index
pred_idx = np.argmax(class_probabilities)
print(f"Predicted Class Index: {pred_idx}")
print(f"Final Model Prediction: {classes[pred_idx]}")
```

### **4. Linear Algebra Core: Matrix Multiplication (`@` / `np.dot`)**

Machine learning me features aur weights ka multiplication normal `*` se nahi, balki **Dot Product / Matrix Multiplication** se hota hai.

* Element-wise multiply: `A * B` (Dono ke shapes exact same chahiye).
* Matrix Multiplication: `A @ B` ya `np.dot(A, B)`.
  - **Rule:** Matrix A ke Columns ($N$) == Matrix B ki Rows ($N$) hone chahiye.
  - Shape Flow: `(M, N) @ (N, P) -> (M, P)`.

```python
# Matrix A: (2 samples, 3 features)
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# Matrix B / Model Weights: (3 features, 2 outputs)
B = np.array([
    [7, 8],
    [9, 1],
    [2, 3]
])

print("Matrix A (2x3):\n", A)
print("\nMatrix B (3x2):\n", B)

# Matrix Multiplication using '@' operator
result = A @ B
print("\nMatrix Multiplication Result (A @ B) -> Shape (2x2):\n", result)
```

### **5. Reproducibility & Shuffling (`seed` & `permutation`)**
* `np.random.seed(number)`: Random generation ko lock karta hai taaki har baar run karne par exact same random numbers aayein (Reproducibility).
* `np.random.permutation(arr)`: Array ka data order randomly shuffle karta hai (Dataset shuffling ke liye).

```python
# 1. Setting Seed for consistent results
np.random.seed(42)
rand_nums = np.random.rand(3)
print("Consistent Random Numbers (Seed 42):", rand_nums)

# 2. Shuffling indices/data
indices = np.arange(6)
shuffled_indices = np.random.permutation(indices)
print("Original Indices:", indices)
print("Shuffled Indices for Train/Test split:", shuffled_indices)
```

### **6. Percentiles & Quartiles (`np.percentile`)**
Data distribution me specific percentage points (25th percentile Q1, 50th median, 75th Q3) nikalne ke liye use hota hai.

```python
dataset = np.array([10, 15, 20, 25, 30, 35, 40, 45, 50, 55, 60])

# 25th (Q1), 50th (Median), 75th (Q3) percentiles
q1 = np.percentile(dataset, 25)
median = np.percentile(dataset, 50)
q3 = np.percentile(dataset, 75)

print(f"25th Percentile (Q1): {q1}")
print(f"50th Percentile (Median): {median}")
print(f"75th Percentile (Q3): {q3}")
print(f"Interquartile Range (IQR = Q3 - Q1): {q3 - q1}")
```

### **7. Matrix Inversion (`np.linalg.inv`)**
Square matrix ka inverse ($A^{-1}$) nikalne ke liye use hota hai.
$$\text{Property: } A @ A^{-1} = I \text{ (Identity Matrix)}$$

```python
square_mat = np.array([[2, 1], 
                       [5, 3]])

# Matrix Inverse
inv_mat = np.linalg.inv(square_mat)
print("Original Matrix (2x2):\n", square_mat)
print("\nInverse Matrix:\n", inv_mat)

# Verification: A @ A_inv should give Identity Matrix
identity_check = np.round(square_mat @ inv_mat)
print("\nA @ Inverse (Identity Matrix):\n", identity_check)
```
