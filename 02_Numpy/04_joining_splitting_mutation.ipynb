# **NumPy Phase 4: Joining, Splitting & Array Mutation**

Is phase me hum seekhenge ki alag-alag arrays ko aapas me kaise jodte hain (Joining), bade array ke tukde kaise karte hain (Splitting), aur array me naye elements add ya delete kaise karte hain (Mutation).

---

### **1. Joining Arrays: `concatenate()` vs `stack()`**

Do ya zyada arrays ko aapas me jodna:

* **`np.concatenate()`:** Existing dimension ke along arrays ko jodta hai (Naya bracket ya dimension add **nahi** hota).
  * `axis=0` $\implies$ Row-wise jodna (Vertical join).
  * `axis=1` $\implies$ Column-wise jodna (Horizontal join).

* **`np.stack()`:** Arrays ko jodte waqt ek **naya dimension (bracket)** create karta hai (e.g., do 1D arrays milkar 2D matrix ban jaate hain).

* **Shortcuts:**
  * **`np.vstack()` (Vertical Stack):** Rows ke roop me ek ke niche ek jodta hai (`axis=0`).
  * **`np.hstack()` (Horizontal Stack):** Columns ke roop me barabar me jodta hai (`axis=1`).

```python
import numpy as np

# Base Arrays
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Array a:", a)
print("Array b:", b)

# 1. concatenate (Result remains 1D)
concat_arr = np.concatenate((a, b))
print("\nconcatenate (1D):", concat_arr)
print("Shape:", concat_arr.shape)

# 2. stack (Creates a new 2D array)
stacked_arr = np.stack((a, b), axis=0)
print("\nstack axis=0 (Creates 2D):\n", stacked_arr)
print("Shape:", stacked_arr.shape)

# 3. vstack & hstack
print("\nvstack (Rows ke roop me):\n", np.vstack((a, b)))
print("\nhstack (Single line columns):\n", np.hstack((a, b)))
```

### **2. Splitting Arrays: `split()`, `hsplit()`, `vsplit()`**

Ek array ko do ya usse zyada barabar hisson me divide karna:

* **`np.split(array, parts, axis)`:** Specified axis par array ke barabar tukde karta hai.
* **`np.vsplit(array, parts)`:** Matrix ko rows ke hisaab se beech me se kat-ta hai (Vertical split).
* **`np.hsplit(array, parts)`:** Matrix ko columns ke hisaab se alag karta hai (Horizontal split).

```python
# 1D Splitting
arr_1d = np.arange(1, 7)  # [1, 2, 3, 4, 5, 6]
parts = np.split(arr_1d, 3)  # 3 equal parts of 2 elements each
print("1D split into 3 parts:")
for p in parts:
    print(p)

print()

# 2D Splitting (Matrix of 4x4)
mat = np.arange(16).reshape(4, 4)
print("Original 4x4 Matrix:\n", mat)

# vsplit (Row-wise cut: 2 matrices of 2x4)
top, bottom = np.vsplit(mat, 2)
print("\nvsplit (Top Half):\n", top)
print("\nvsplit (Bottom Half):\n", bottom)

# hsplit (Column-wise cut: 2 matrices of 4x2)
left, right = np.hsplit(mat, 2)
print("\nhsplit (Left Half):\n", left)
print("\nhsplit (Right Half):\n", right)
```

### **3. Array Mutation: `append()`, `insert()`, `delete()`**

NumPy array me naye elements jodne ya purane elements hatane ke liye:

* **`np.append(array, values)`:** Array ke aakhri (end) me nayi values jodta hai.
* **`np.insert(array, index, values)`:** Kisi specific index par element insert karta hai.
* **`np.delete(array, index)`:** Kisi specific index ke element ya poori row/column ko delete karta hai.

> **Important Note:** NumPy array ka memory size fixed hota hai. Isliye jab bhi hum `append` ya `insert` karte hain, NumPy internally ek **naya array** banakar return karta hai.

```python
x = np.array([10, 20, 30, 40, 50])
print("Original array:", x)

# 1. append
appended = np.append(x, [60, 70])
print("\nAfter np.append:", appended)

# 2. insert (Index 2 par 99 add karna)
inserted = np.insert(x, 2, 99)
print("\nAfter np.insert at index 2:", inserted)

# 3. delete (Index 1 ki value 20 delete karna)
deleted = np.delete(x, 1)
print("\nAfter np.delete index 1:", deleted)

# 2D Matrix row/column delete
grid = np.array([[1, 2], [3, 4], [5, 6]])
print("\nOriginal 2D Grid:\n", grid)

# Row delete (axis=0) -> Delete row index 1
grid_no_row = np.delete(grid, 1, axis=0)
print("\nGrid after deleting Row 1 (axis=0):\n", grid_no_row)
```

### **4. Advanced Array Mutation & Replacement Operations**

Jab array ke shape ko delete kiye bina sirf uske andar ki values ko modify ya transform karna ho:

---

#### **1. `np.clip()` (Values ko Limit / Cap karna)**
Kisi array ki values ko ek minimum aur maximum boundary ke andar lock kar deta hai.
* Agar koi value `min` se choti hai $\to$ Use `min` bana deta hai.
* Agar koi value `max` se badi hai $\to$ Use `max` bana deta hai.
* **Use-Case:** Outliers ko trim karna ya pixel values ko 0 se 255 ke beech clamp karna.

**Syntax:** `np.clip(array, min_value, max_value)`

---

#### **2. `np.repeat()` vs `np.tile()` (Elements ko Multiply karna)**
* **`np.repeat(arr, repeats)`:** Array ke har ek individual element ko baar-baar repeat karta hai (e.g., `[1, 2]` $\to$ `[1, 1, 2, 2]`).
* **`np.tile(arr, reps)`:** Poore array ke block ko ek saath repeat karta hai (e.g., `[1, 2]` $\to$ `[1, 2, 1, 2]`).

---

#### **3. `np.unique()` (Duplicates hatana & Frequency nikalna)**
Array me se repeated values hata kar unique sorted values deta hai aur unka count bhi return kar sakta hai.

**Syntax:** `np.unique(array, return_counts=True)`

```python
# 1. np.clip (Outlier Capping)
raw_scores = np.array([-15, 20, 55, 90, 140, 200])
print("Original Scores:", raw_scores)

# Values ko 0 aur 100 ke beech cap karna
clean_scores = np.clip(raw_scores, 0, 100)
print("Clipped Scores (0 to 100):", clean_scores)

print()

# 2. np.repeat vs np.tile
sample = np.array([1, 2, 3])
print("Original Sample:", sample)

# Element-wise repeat
print("np.repeat (har number 3 baar):", np.repeat(sample, 3))

# Full pattern tile
print("np.tile (poora pattern 3 baar):", np.tile(sample, 3))

print()

# 3. np.unique (Duplicates & Counts)
data = np.array([5, 2, 8, 2, 5, 5, 9, 8])
unique_vals, counts = np.unique(data, return_counts=True)

print("Original Array with Duplicates:", data)
print("Unique Elements:", unique_vals)
print("Har Element ka Count:", counts)
```
